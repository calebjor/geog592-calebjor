# Cleaning U.S. County Population Estimates

This notebook cleans a U.S. Census Bureau CSV containing annual county population estimates from 2020 through 2025. The goal is to create a simpler dataset that contains county identification information and a recent population estimate that could be joined to a county vector dataset in GIS.

```python
import csv 

filename = "data/GEOG592-HW3-USAnnualEstPop.csv"

with open(filename, "r", encoding="utf-8-sig") as file:
    reader = csv.reader(file)
    rows = list(reader)

for row in rows[:10]:
    print(row)
```

## 2. Inspect and Filter the Dataset

The CSV contains several header rows at the top, a United States total row, and source notes at the bottom. By inspecting the file, I found that county-level records begin with a leading period (`.`) in the Geographic Area field. I used that pattern to keep only county and county-equivalent rows while removing the extra header and footer information.

```python
print("Total rows:", len(rows))

county_rows = [
    row for row in rows
    if len(row) >= 8 and row[0].startswith(".")]

print("\nCounty rows:", len(county_rows))

for row in county_rows[:10]:
    print(row)
```

## 3. Select and Clean the Needed Fields

For the cleaned dataset, I kept the Geographic Area field and the annual population estimates from 2020 through 2025. I removed the leading period from the Geographic Area field and removed commas from the population values so they could be stored as integers. The Geographic Area field can potentially be used to join the CSV to a county vector dataset.

```python
cleaned_data = []

for row in county_rows:
    geographic_area = row[0][1:] # removes leading dot

    pop_2020 = int(row[2].replace(",", ""))
    pop_2021 = int(row[3].replace(",", ""))
    pop_2022 = int(row[4].replace(",", ""))
    pop_2023 = int(row[5].replace(",", ""))
    pop_2024 = int(row[6].replace(",", ""))
    pop_2025 = int(row[7].replace(",", ""))

    cleaned_data.append([geographic_area, pop_2020, pop_2021, pop_2022, pop_2023, pop_2024, pop_2025])

for row in cleaned_data[:10]:
    print(row)
```

## 5. Export the Cleaned CSV

The cleaned data is exported to a new CSV file in the `data` folder. The output includes the Geographic Area field and annual population estimates from 2020 through 2025.

```python
output_file = "data/GEOG592-HW3-USAnnualEstPop-cleaned.csv"

with open(output_file, "w", newline="", encoding="utf-8-sig") as file:
    writer = csv.writer(file)

    writer.writerow([
        "geographic_area",
        "population_2020",
        "population_2021",
        "population_2022",
        "population_2023",
        "population_2024",
        "population_2025"])

    for row in cleaned_data:
         writer.writerow(row)

print("Cleaned CSV saved as:", output_file)
```
