**Question 23**

Write code to calculate the mean and median of `Visitors`

```python
parks = pd.DataFrame({
    "Park": ["Oak", "Pine", "Maple", "Cedar"],
    "Area": [12, 25, 18, 40],
    "Visitors": [120, 250, 180, 310]
})


print("Mean:", parks["Visitors"].mean())

print("Median:", parks["Visitors"].median())
```
