## 2. Filter the Data w/ a Boolean Mask

I created a Boolean mask using the 2025 population column. The mask identifies counties with a 2025 population greater than 500,000. I then used the mask to create a smaller DataFrame containing only those counties.

```python
large_pop = df["population_2025"] > 500000

large_pop.head()
```

```python
large_counties = df[large_pop]

large_counties.head()

print("Original df:", df.shape)
print("Filtered df:", large_counties.shape)
```
