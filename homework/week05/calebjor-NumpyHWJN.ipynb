**Boolean Masks & Spatial Queries**

```python
x = points[:, 0]
y = points[:, 1]

inside = (x >= 110) & (x <= 150) & (y >= 220) & (y <= 290)
```

```python
# Write the boolean values stored inside [:]
 # - [False, True, False, True]
inside[:]
```

```python
# points[inside]? 
 # - array([[140,230], [120, 280]])
points[inside]
```

```python
# In GIS language, what spatial operation is this Boolean mask approximating?
 # - It is approximating points that fall w/in a rectangular bounding (box) area
```

**np.where() & Classification**

```python
elevation = np.array([240, 510, 430, 680])
classes = np.where(elevation >= 500, 1, 0)
```

```python
# What values are stored in classes[:]?
 # - [0, 1, 0, 1]
classes[:]
```

```python
# Explain the 3 arguments in np.where(condition, value_if_true, value_if_false)
 # - condition is the test being applied to the array; value_if_true is the value assigned to the array if true; value_if_false is the value assigned to the array if false
```

**Big Picture**

| Structure | Expression | What it selects |
|---|---|---|
| Python list | `locations[2]` | **The 3rd item in the list** |
| Nested Python list | `points_list[2][1]` | **The 2nd item from the 3rd inner list** |
| pandas DataFrame | `df.iloc[2, 1]` | **The value in the 3rd row and 2nd column** |
| NumPy array | `points[2, 1]` | **The value in the 3rd row and 2nd column (y-coord of 3rd point)** |
| NumPy array | `points[:, 0]` | **All rows from the 1st column (All x-coords)** |
