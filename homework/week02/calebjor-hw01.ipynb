# Part 4: Calculate the Distance from A to B

Create a function named `calculate_distance` that has two parameters:

```python
calculate_distance(point1, point2)
```

The function should calculate the straight-line distance between two projected coordinates:

distance(A,B) $= \sqrt{(A_\text{easting} - B_\text{easting})^2 +  (A_\text{northing} - B_\text{northing})^2}$

You may want to use the python exponent operator `**`, and to remember that the square root of a number is the number to the power $\frac{1}{2}:$ 


```python
A_squared = A**2 #this is the square of A
A_sqrt = A**0.5 #this is the square root of A  
```

The function should **return** the calculated distance.

Use the function to calculate the distance between A and B, and *print* the result.

````python
def calculate_distance(point2, point1):
    eastings_diff = point1[0] - point2[0]
    northings_diff = point1[1] - point2[1]
    distance = (eastings_diff**2 + northings_diff**2)**0.5
    return distance

distance_AB = calculate_distance(point_A, point_B)
print(distance_AB)
````

# Part 5: Check the Input Data

Now improve `calculate_distance()` so that it verifies the input data before calculating the distance.

Your function should check that:

1. `point1` is a list;
2. `point2` is a list;
3. each coordinate list contains exactly two values;
4. every coordinate value has a datatype of either `int` or `float`.

For example, your function should recognize these as numeric values:

```python
type(680100) == int
type(680100.5) == float
```

If the input is not valid:

- print a useful error message;
- use `return` to stop the function before attempting the distance calculation.

````python
def calculate_distance(point1, point2):

    if type(point1) != list:
        print("Error: point1 is not a list.")
        return
    if type(point2) != list:
        print("Error: point2 is not a list.")
        return

    if len(point1) != 2:
        print("Error: point1 must contain exactly two values.")
        return
    if len(point2) != 2:
        print("Error: point2 must contain exactly two values.")
        return

    for value in point1:
        if type(value) != int and type(value) != float:
            print("Error: point1 contains a non-numeric value.")
            return
    for value in point2:
        if type(value) != int and type(value) != float:
            print("Error: point2 contains a non-numeric value.")
            return

    eastings_diff = point1[0] - point2[0]
    northings_diff = point1[1] - point2[1]
    distance = (eastings_diff**2 + northings_diff**2)**0.5
    return distance

distance_AB = calculate_distance(point_A, point_B)
print(distance_AB)
````
